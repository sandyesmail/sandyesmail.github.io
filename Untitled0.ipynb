```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.linspace(0,10,100)
y=x**2
fig, ax = plt.subplots()
ax.plot (x,y)
ax.set_xlabel("x")
ax.set_ylabel("y=^2")
plt.show()
```

```python
y1 = x**2
y2 = x**3

plt.plot(x,y1, label='y=x^2')
plt.plot(x,y2, label='y=x^3')
plt.title('Plot of y = x^2 and y=x^3')
plt.legend()
plt.grid(True)

plt.show()
```
