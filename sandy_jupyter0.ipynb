```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.linspace(0,10,100)
y=x**2
fig, ax = plt.subplots()
ax.plot (x,y)
ax.set_xlabel("x")
ax.set_ylabel("y=^2")
plt.show()
```

```python
y1 = x**2
y2 = x**3

plt.plot(x,y1, label='y=x^2')
plt.plot(x,y2, label='y=x^3')
plt.title('Plot of y = x^2 and y=x^3')
plt.legend()
plt.grid(True)

plt.show()
```

```python
chrom_data=pd.read_excel('/content/chrosome_basepairs.xlsx')
```

```python
chrom_data
```

```python
bp_length = 3.4
chrom_data["Length (Angstrom)"] = chrom_data["basepais(bp)"] * bp_length
chrom_data.head()
```

```python
total_length_angstrom = chrom_data["Length (Angstrom)"].sum()
total_length_meter = total_length_angstrom / 1e10
print(f"Total length of all chromosomes in meters: {total_length_meter} m")
```

```python
import pandas as pd

# Total base pairs in the human genome
total_basepairs = 3_070_536_000   # 3.07 billion
total_mbps = total_basepairs / 1_000_000   # convert to megabase pairs

# Cost data for each year
cost_data = {
    'Year': [2001, 2009, 2010, 2020],
    'Cost per MBps ($)': [10000, 1.00, 0.01, 0.001]
}

# Create DataFrame
cost_df = pd.DataFrame(cost_data)

# Calculate total sequencing cost
cost_df['Total Sequencing Cost ($)'] = cost_df['Cost per MBps ($)'] * total_mbps

# Display the table
cost_df
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

sns.barplot(
    x='Year',
    y='Total Sequencing Cost ($)',
    data=cost_df,
    palette='viridis'
)

plt.yscale('log')
plt.xlabel('Year')
plt.ylabel('Total Sequencing Cost ($)')
plt.title('Total Whole Genome Sequencing Cost by Year')

plt.show()
```
